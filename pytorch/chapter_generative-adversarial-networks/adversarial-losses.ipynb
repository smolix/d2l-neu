# Adversarial Losses Beyond GANs

When predicting a single scalar such as a house price, minimizing mean squared error ($L_2$ loss) is standard practice. On conditional image generation problems, such as super-resolution (mapping a low-resolution thumbnail $c$ to a high-resolution image $x$) or sketch-to-photo translation (pix2pix, @Isola.Zhu.Zhou.ea.2017), pixelwise $L_2$ and $L_1$ regression losses produce blurry outputs. Many distinct high-resolution textures correspond to the same low-resolution input, and minimizing pixelwise squared error forces the network to output their conditional average $\mathbb{E}[x \mid c]$, a blur that belongs to none of the plausible sharp images. Pairing the regression loss with a discriminator $D(x, c)$ that scores whether the output is a realistic image for condition $c$ restores sharp high-frequency detail.

This mechanism explains why adversarial losses remain widely used even though diffusion and flow-matching models (that section) have replaced standalone GANs for general image synthesis. A learned discriminator still trains one-step generators distilled from multi-step diffusion models, the image autoencoders (tokenizers) on which latent diffusion models operate, and neural audio vocoders that reconstruct speech waveforms from spectrograms. Below, we analyze the conditional-mean averaging effect on a minimal bimodal regression problem, demonstrate why one-step distillation from a high-noise denoising teacher collapses to the conditional mean without a discriminator, and summarize where adversarial losses are used in modern generative pipelines.

```python
import math
import numpy as np
import torch
from torch import nn
from torch.nn import functional as F
from d2l import torch as d2l
```

## Conditional Means and Multimodal Targets

### The Conditional-Mean Decomposition

Let $g(c)$ be a deterministic predictor of $x$ given condition $c$, and let $\mu(c) = \mathbb{E}[x \mid c]$ denote the conditional mean. Adding and subtracting $\mu(c)$ inside $\mathbb{E}[\|x - g(c)\|^2]$ eliminates the cross term $2\,\mathbb{E}[\langle x - \mu(c), \mu(c) - g(c) \rangle] = 0$ (by iterated expectation given $c$), leaving the orthogonal decomposition:

$$
\begin{aligned}
\mathbb{E}\big[(x - g(c))^2\big]
&= \mathbb{E}\big[(x - \mu(c))^2\big] + \mathbb{E}\big[(\mu(c) - g(c))^2\big].
\end{aligned}
$$

The first term is the irreducible conditional variance $\operatorname{Var}(x \mid c)$, and the second is the squared distance between $g(c)$ and $\mu(c)$. Thus, squared error is uniquely minimized by the conditional mean $g^\star(c) = \mu(c)$. When $p(x \mid c)$ is unimodal and concentrated, $\mu(c)$ is a representative target; when $p(x \mid c)$ is multimodal, $\mu(c)$ falls in the low-density region between modes. Replacing $L_2$ with $L_1$ loss $\mathbb{E}[|x - g(c)|]$ merely replaces the conditional mean with the coordinate-wise conditional median (Exercise 2), which likewise lies outside the support of a symmetric bimodal distribution.

### A Bimodal Two-Band Benchmark

Consider a 1D conditional distribution with $c \sim \textrm{Uniform}[-1, 1]$ and two parallel data bands separated by an empty corridor:

$$
\begin{aligned}
x &= c + b + 0.05\,\xi, \\
b &\in \{-1, +1\} \textrm{ with probability } \tfrac12 \textrm{ each},
\quad \xi \sim \mathcal{N}(0, 1).
\end{aligned}
$$

The conditional mean $\mu(c) = c$ runs down the empty corridor between the bands $x = c \pm 1$, achieving $\mathbb{E}[(x - \mu(c))^2] = 1 + 0.05^2 = 1.0025$. By contrast, a predictor that commits to the upper band $g(c) = c + 1$ has squared error $\tfrac12(0) + \tfrac12(2^2) + 0.05^2 = 2.0025$—twice as large. Squared error therefore penalizes committing to a valid data mode.

A conditional discriminator $D(x, c)$, however, estimates the log density ratio (that section):

$$
D^\star(x, c) = \log \frac{p(x, c)}{q(x, c)}.
$$

In the empty corridor where $p(x, c) \approx 0$, $D^\star(x, c)$ is strongly negative, so minimizing the non-saturating adversarial loss $-\log \sigma(D(\hat x, c))$ forces the predictor $\hat x = g(c)$ onto one of the high-density data bands.

We test this by training two identical 2-hidden-layer $\tanh$ MLPs (`g_mse` and `g_adv`) from the same initial weights: `g_mse` minimizes squared error, while `g_adv` trains against a leaky-ReLU discriminator $D(x, c)$ using `d2l.gan_loss_D` and `d2l.gan_loss_G` (that section).

```python
def make_bands(n):
    """Draw n pairs (c, x) from the two bands x = c + b + 0.05 xi."""
    c = torch.rand(n, 1) * 2 - 1
    b = torch.randint(0, 2, (n, 1)) * 2.0 - 1.0
    return c, c + b + 0.05 * torch.randn(n, 1)

torch.manual_seed(0)
g_mse = d2l.make_mlp([1, 32, 32, 1], nn.Tanh())
g_adv = d2l.make_mlp([1, 32, 32, 1], nn.Tanh())
g_adv.load_state_dict(g_mse.state_dict())  # both students start alike
D = d2l.make_mlp([2, 32, 32, 1], nn.LeakyReLU(0.2))
c_data, x_data = make_bands(200)  # a data sample for the plot
```

The first student minimizes squared error with Adam, its learning rate decaying along a cosine from 0.01 to zero over 2,000 steps. We need no experiment to know where it ends: by the equation it should follow the line $x = c$, one unit from each band. It is the control for the second student.

```python
def train_mse(g, num_steps=2000, batch_size=128, lr=0.01):
    opt = torch.optim.Adam(g.parameters(), lr=lr)
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, num_steps)
    for _ in range(num_steps):
        c, x = make_bands(batch_size)  # a fresh batch at every step
        opt.zero_grad()
        ((g(c) - x) ** 2).mean().backward()
        opt.step()
        sched.step()

train_mse(g_mse)
```

The second student plays the log-loss game, with the losses `d2l.gan_loss_D` and `d2l.gan_loss_G` of that section. At each step the critic takes one logistic-regression step on real pairs $(x, c)$ against predicted pairs $(\hat x, c)$, and then the student takes one step on $-\log \sigma\big(D(\hat x, c)\big)$ against the updated critic. The prediction enters the game with the data's observation noise, $\hat x = g(c) + 0.05\,\xi'$ with a fresh $\xi' \sim \mathcal{N}(0, 1)$, so the two distributions differ in where they put their mass, not in how smooth they are. A game moves more slowly than a regression, so this student gets 8,000 steps; the squared-error student is at its optimum after 2,000, and more steps would not move it.

```python
def train_adversarial(g, D, num_steps=8000, batch_size=128):
    opt_G = torch.optim.Adam(g.parameters(), lr=1e-3)
    opt_D = torch.optim.Adam(D.parameters(), lr=2e-3)
    for _ in range(num_steps):
        c, x = make_bands(batch_size)
        x_hat = g(c) + 0.05 * torch.randn(batch_size, 1)  # the data's noise
        real, fake = torch.cat([c, x], 1), torch.cat([c, x_hat], 1)
        opt_D.zero_grad()
        d2l.gan_loss_D(D, real, fake.detach()).backward()
        opt_D.step()
        opt_G.zero_grad()
        d2l.gan_loss_G(D, real, fake).backward()
        opt_G.step()

train_adversarial(g_adv, D)
```

Three numbers summarize each student on a grid of 201 inputs: the mean distance from its curve to the nearer band, the mean distance to the conditional mean $x = c$, and the fraction of inputs at which the curve is more than 0.25 from both bands.

```python
cs = torch.linspace(-1, 1, 201).reshape(-1, 1)
with torch.no_grad():
    curves = {'squared error': g_mse(cs), 'adversarial': g_adv(cs)}
for name, x_hat in curves.items():
    to_band = torch.minimum((x_hat - cs - 1).abs(), (x_hat - cs + 1).abs())
    print(f'{name}: {to_band.mean():.2f} from the nearer band, '
          f'{(x_hat - cs).abs().mean():.2f} from x = c, '
          f'{(to_band > 0.25).float().mean():.0%} off the bands')
```

Plotting both learned curves in `curves` over the training sample
`(c_data, x_data)` alongside the dashed conditional mean $x = c$ makes the
difference between regression and adversarial training immediately visible:

```python
d2l.plot([c_data, cs, cs, cs],
         [x_data, curves['squared error'], curves['adversarial'], cs],
         '$c$', '$x$', ['data', 'squared-error student',
                        'adversarial student', '$x = c$'],
         ylim=(-2.2, 3.6), fmts=('C0.', 'C2-', 'C1-', 'k--'),
         figsize=(5.5, 3.2))
```

As predicted, the squared-error model `g_mse` tracks the conditional mean $\mu(c) = c$ down the empty corridor (about one unit from either band, with $100\%$ of predictions off the bands). The adversarial model `g_adv` commits to valid data bands (a few hundredths from the nearest band on average, with at most $2\%$ of points crossing the corridor).

Notice the fundamental perception–distortion trade-off [@Blau.Michaeli.2017]: even an ideal stochastic sampler $\hat x \sim p(x \mid c)$ has expected squared error $2\,\operatorname{Var}(x \mid c)$—twice the MSE (or $3\text{ dB}$ lower PSNR) of the blurry conditional mean $\mu(c)$ (Exercise 1). Furthermore, because a deterministic function $g(c)$ can output only one value per input $c$, covering *both* bands simultaneously requires passing a stochastic latent variable $z$ alongside $c$: $\hat x = g(c, z)$ (Exercise 3).

## Discriminators in Diffusion Distillation

### Denoisers as Conditional Means

Diffusion and flow-matching models (that section) train a denoiser network by squared error to predict clean data $x_0$ from a noisy input $x_\tau = x_0 + \tau\xi$ ($\xi \sim \mathcal{N}(0, I)$). By the equation, the optimal denoiser computes the posterior conditional mean $\mathbb{E}[x_0 \mid x_\tau]$ (that section). At large noise levels $\tau$, many clean images are consistent with $x_\tau$, so $\mathbb{E}[x_0 \mid x_\tau]$ is a blurry average; multi-step samplers avoid returning this average by taking many small steps from high noise down to $\tau \to 0$.

When *distilling* a multi-step diffusion teacher into a fast one-step student generator (that section), one common approach adds noise $\tau\xi$ to the student's output $\hat x$ and regresses $\hat x$ toward the frozen teacher's denoised estimate $\mathbb{E}[x_0 \mid \hat x + \tau\xi]$. At high noise $\tau$, however, the teacher's denoised target is itself a conditional mean—pulling the one-step student back toward the blurry average unless an adversarial critic is included.

### Closed-Form Distillation on the Two-Band Benchmark

On the two-band problem the equation, the exact Bayes-optimal denoising teacher at noise level $\tau$ has a closed form. Given a noisy observation $\tilde x = x + \tau\xi$ of a point on band $b \in \{-1, +1\}$, $\tilde x \mid c, b \sim \mathcal{N}(c + b, \tau^2 + 0.05^2)$. By Bayes' rule, the posterior log-odds of $b = +1$ versus $b = -1$ are $2(\tilde x - c)/(\tau^2 + 0.05^2)$, giving the posterior mean band center:

$$
\mathbb{E}[c + b \mid \tilde x, c] \;=\; c + \tanh\Big(\frac{\tilde x - c}{\tau^2 + 0.05^2}\Big).
$$

We now equip each student $g(c, z)$ with a 1D Gaussian latent $z \sim \mathcal{N}(0, 1)$ so it has the capacity to cover both bands, and compare three training objectives at teacher noise level $\tau = 2$:

1. *Regression only*: minimizing $\|\hat x - \operatorname{stop\_grad}(\text{teacher}(\hat x + \tau\xi, c, \tau))\|^2$. Because the slope of the equation in $\tilde x$ is at most $(\tau^2 + 0.05^2)^{-1} \approx \tfrac14 < 1$ at $\tau = 2$, the fixed-point equation $u = \mathbb{E}_\xi[\tanh((u + \sqrt{\tau^2 + 0.05^2}\,\xi)/(\tau^2 + 0.05^2))]$ for the offset $u = g(c, z) - c$ has the unique contraction fixed point $u = 0$ (the empty corridor).
2. *Critic only*: training $g(c, z)$ against a conditional discriminator $D(x, c)$ with `d2l.gan_loss_D` and `d2l.gan_loss_G`.
3. *Critic + regression*: summing the adversarial loss and the teacher regression loss with equal weight (as in Adversarial Diffusion Distillation, ADD, @Sauer.Lorenz.Blattmann.ea.2023).

```python
def teacher(x_tilde, c, tau):
    """Posterior mean of the band center c + b given x_tilde = x + tau xi."""
    return c + torch.tanh((x_tilde - c) / (tau ** 2 + 0.05 ** 2))

def train_student(g, D, lam, tau=2.0, num_steps=4000, batch_size=128):
    """Train g(c, z) on the critic's loss if D is given, plus lam times the
    squared error toward the teacher's estimate at noise level tau."""
    opt_G = torch.optim.Adam(g.parameters(), lr=1e-3)
    if D is not None:
        opt_D = torch.optim.Adam(D.parameters(), lr=2e-3)
    for _ in range(num_steps):
        c, x = make_bands(batch_size)
        z = torch.randn(batch_size, 1)  # the student's latent
        x_hat = g(torch.cat([c, z], 1)) + 0.05 * torch.randn(batch_size, 1)
        real, fake = torch.cat([c, x], 1), torch.cat([c, x_hat], 1)
        loss_G = 0
        if D is not None:
            opt_D.zero_grad()
            d2l.gan_loss_D(D, real, fake.detach()).backward()
            opt_D.step()
            loss_G = d2l.gan_loss_G(D, real, fake)
        with torch.no_grad():  # the teacher's estimate is a fixed target
            target = teacher(x_hat + tau * torch.randn(batch_size, 1), c, tau)
        loss_G = loss_G + lam * ((x_hat - target) ** 2).mean()
        opt_G.zero_grad()
        loss_G.backward()
        opt_G.step()
```

A student is now a distribution of outputs at each $c$. The function `band_shares` evaluates it at the 201 inputs of the first experiment and at 64 evenly spaced quantiles of the latent, and returns the mean distance to the nearer band and the shares of outputs within 0.25 of the upper and of the lower band.

```python
def band_shares(g, cs):
    """Mean distance of g(c, z) to the nearer band, and its shares on the
    upper and the lower band, over the inputs cs and 64 quantiles z."""
    zs = torch.special.ndtri((torch.arange(64) + 0.5) / 64)
    c, z = (v.reshape(-1, 1) for v in torch.meshgrid(
        cs.flatten(), zs, indexing='ij'))
    with torch.no_grad():
        u = g(torch.cat([c, z], 1)) - c  # offset from the corridor
    to_band = torch.minimum((u - 1).abs(), (u + 1).abs())
    return (to_band.mean(), ((u - 1).abs() < 0.25).float().mean(),
            ((u + 1).abs() < 0.25).float().mean())
```

Each student trains from three seeds. Within a seed, the three students start from the same weights and see the same data.

```python
students = {'regression': (False, 1.0), 'critic': (True, 0.0),
            'critic + regression': (True, 1.0)}
for name, (has_critic, lam) in students.items():
    for seed in range(3):
        torch.manual_seed(seed)
        g = d2l.make_mlp([2, 32, 32, 1], nn.Tanh())
        D = d2l.make_mlp([2, 32, 32, 1], nn.LeakyReLU(0.2))
        train_student(g, D if has_critic else None, lam)
        dist, upper, lower = band_shares(g, cs)
        print(f'{name}, seed {seed}: {dist:.2f} from the nearer band, '
              f'{upper:.0%} upper, {lower:.0%} lower, '
              f'{1 - upper - lower:.0%} off the bands')
```

The three-seed results confirm the fixed-point analysis:

- *Regression onto the high-noise teacher* collapses $100\%$ of outputs into the empty corridor (about one unit from either band), ignoring the latent $z$.
- *Critic only* and *Critic + regression* both place $98\text{--}99\%$ of outputs directly on the two data bands (mean distance of a few hundredths, always below $0.1$), splitting probability mass evenly ($\sim 49\% / 49\%$) between the upper and lower bands via the latent $z$. Because the bands have narrow width $\sigma = 0.05$, the optimal critic's restoring slope ($\delta / 0.05^2 = 400\delta$) easily dominates the gentle inward pull of the high-noise teacher.

### Distillation, Autoencoders, and Vocoders at Scale

This combination of a regression or perceptual anchor with an adversarial discriminator appears across three major classes of modern generative systems:

1. **Adversarial Diffusion Distillation (ADD and DMD2):** ADD [@Sauer.Lorenz.Blattmann.ea.2023] trains a one-step image generator by combining a denoising teacher regression loss with a discriminator built on frozen DINOv2 features, finding the same ordering as our experiment above (teacher regression alone produces blurry images, while adding the discriminator yields sharp one-step samples). Distribution Matching Distillation (DMD, @Yin.Gharbi.Zhang.ea.2024) uses the difference between a frozen teacher's score $\nabla_x \log p_\tau$ and a fake-distribution diffusion model's score $\nabla_x \log q_\tau$ to descend $D_{\textrm{KL}}(q_\tau \,\|\, p_\tau)$. We have met this difference before: $\nabla_x \log p_\tau - \nabla_x \log q_\tau = \nabla_x \lambda_\tau$ is the input gradient of the optimal critic $\lambda_\tau = \log(p_\tau/q_\tau)$ between the noised distributions. DMD's generator update is therefore the unweighted ($w = 1$) reverse-KL update of Exercise 5, with the critic's slope supplied by two score networks instead of a discriminator. Since both distributions are noised, their supports overlap at every noise level $\tau > 0$, which is the instance noise of that section. DMD2 [@Yin.Gharbi.Park.ea.2024] adds an adversarial discriminator against real images so the one-step student can exceed the teacher's quality.
2. **Latent Image Autoencoders (Tokenizers):** Latent diffusion models (that section) compress images into a lower-dimensional latent space using an autoencoder (such as VQGAN or the Stable Diffusion VAE, @Esser.Rombach.Ommer.2021; @Rombach.Blattmann.Lorenz.ea.2022). Because the compressed latent code leaves fine-grained local texture underdetermined, training the decoder with pixelwise $L_1/L_2$ and perceptual (LPIPS) losses is augmented with a patch-level discriminator (PatchGAN) to ensure crisp high-frequency textures.
3. **Neural Speech Vocoders:** Text-to-speech pipelines typically predict a mel-spectrogram (which specifies frame-level frequency magnitudes but discards phase) and use a vocoder such as HiFi-GAN [@Kong.Kim.Bae.2020] to invert the spectrogram into an audio waveform. Because the expected value of a sinusoid $\sin(\omega n + \vartheta)$ over an unknown phase $\vartheta \sim \textrm{Uniform}[-a, a]$ shrinks in amplitude by $\sin(a)/a$ (vanishing completely for $\vartheta \sim \textrm{Uniform}[-\pi, \pi]$, Exercise 5), waveform regression collapses to silence; HiFi-GAN instead pairs a mel-spectrogram $L_1$ loss and a feature-matching loss on discriminator activations with multi-period and multi-scale waveform discriminators.

## Summary

Minimizing squared error $\mathbb{E}[\|x - g(c)\|^2]$ drives a predictor toward the conditional mean $\mu(c) = \mathbb{E}[x \mid c]$ (the equation), which averages across distinct modes whenever the target $x$ is ambiguous given $c$. Pairing the predictor with a conditional discriminator $D(x, c)$ that estimates $\log(p(x, c)/q(x, c))$ penalizes predictions in low-density corridors between modes and forces outputs onto the data manifold, trading up to a twofold increase in MSE ($3\text{ dB}$ in PSNR) for perceptual realism [@Blau.Michaeli.2017]. Because a diffusion denoiser likewise computes the posterior mean $\mathbb{E}[x_0 \mid x_\tau]$, distilling a one-step generator by regressing directly onto a high-noise denoiser contracts predictions toward the blurry average unless an adversarial critic is included. For the same reason, adversarial discriminators remain essential components of latent image autoencoders, neural speech vocoders, and one-step distilled diffusion generators.
